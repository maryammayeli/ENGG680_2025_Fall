```python
plt.figure(figsize=(10, 6))
plt.hist(df_studentGrades['mathGrade'], bins=10, alpha=0.7, lable='Math Grade')
plt.xlable('Grade')
plt.ylable('Frequency')
plt.title('Distribution of Math Grades')
plt.legend(loc='upper right')
plt.grid(True)
plt.savefig('hist_math.png')
plt.show
```

```python
plt.figure(figsize=(10, 6))
plt.scatter(df_studentGrades['englishGrade'], df_studentGrades['mathGrade'])
plt.xlable('English Grade')
plt.ylable('Math Grade')
plt.title('English Grade vs Math Grade')
plt.grid(True)
plt.savefig('scatter_english_math.png')
plt.show
```

```python
corr = df_studentGrades['englishGrade'].corr(df_studentGrades['mathGrade'])
print(f'Correlation (English vs Math): {corr:.2f}')

# Simple automatic interpretation text:
if corr >= 0.7:
    level = "strong positive"
elif corr >= 0.4:
    level = "moderate positive"
elif corr >= 0.2:
    level = "weak positive"
elif corr <= -0.7:
    level = "strong negative"
elif corr <= -0.4:
    level = "moderate negative"
elif corr <= -0.2:
    level = "weak negative"
else:
    level = "little to no"

print(f"Interpretation: There is a {level} relationship between English and Math grades.")
print("Visual check: points should generally rise from left-bottom to right-top if positive.")
```
